```python
import matplotlib.pyplot as plt
name=data["Name"]
score=data['Exam_Score']
plt.bar(name, score)
plt.title('Student exam scores')
plt.xlabel('Student name')
plt.ylabel('Exam score')
plt.show()
```

```python
result=data['Result'].value_counts()
plt.pie(result,labels=result.index,autopct='%1.0f%%')
plt.title('result distribution')
plt.show()
```

```python
stundent_id=data['Student_ID']
exam_score=data['Exam_Score']
plt.scatter(stundent_id, exam_score)
plt.title('Student ID vs Exam Score')
plt.xlabel('Student ID')
plt.ylabel('Exam Score')
plt.show()
```
