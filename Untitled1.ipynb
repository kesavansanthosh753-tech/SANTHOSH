```python
import matplotlib.pyplot as plt
name = data["day"]
score = data['temp_low']
plt.bar(name, score)
plt.title('Daily low Temperatures')
plt.xlabel('Day')
plt.ylabel('Temperature low (°C)')
plt.show()
```

```python
result=data['condition'].value_counts()
plt.pie(result,labels=result.index,autopct='%1.0f%%')
plt.title('Weather Condition Distribution')
plt.show()
```
